# OpenCode proxy — minimal limit switch

One cell. It starts the proxy and the zrok address, prints every proxy log line, and stays running.

On a free-limit response (`upstream 429` / `FreeUsageLimitError`) it posts `oc-limit-hit` to the Tampermonkey script and calls `runtime.unassign()`. The script reloads after the session is gone and presses Ctrl+F9, which runs this cell again.

Secrets (🔑, Notebook access): `ZROK_TOKEN`, `ZROK_NAME`. Add `OC_PROXY_GITHUB_REPO_TOKEN` only if the repo is private.

Leave the cell on `[*]`.

```python
# Leave this cell running. Colab runs it on startup.
import urllib.request
try:
    from google.colab import userdata
    try:
        _tok = (userdata.get('OC_PROXY_GITHUB_REPO_TOKEN') or '').strip()
    except Exception:
        _tok = (userdata.get('GITHUB_TOKEN') or '').strip()
except Exception:
    _tok = ''
_req = urllib.request.Request(
    'https://api.github.com/repos/recursive-factory/opencode-proxy/contents/notebooks/colab_min.py',
    headers={'Accept': 'application/vnd.github.raw'})
if _tok:
    _req.add_header('Authorization', 'Bearer ' + _tok)
exec(urllib.request.urlopen(_req).read().decode())
```
